```python
#Cleaning the dataset
df1 = df
df1.head()
```

```python
df1 = df1.dropna()
```

```python
df1.shape
```

```python
df1.info()
```

```python
df1 = df1.drop(columns = ['review_title','review_id','review_content','img_link','product_link'])
```

```python
df1 = df1.drop(columns = ['about_product'])
```

```python
df1.shape
```

```python
df1.info()
```

```python
df1.head()
```

```python
#Converting the datatypeof rating from str to float

df1["rating"] = df1["rating"].str.replace("|", "0")
df1['rating'] = df1['rating'].astype(float)

```

```python
#Converting the datatypeof rating_count from str to float
df1['rating_count'].value_counts()
```

```python
df1['rating_count'] = df1['rating_count'].str.replace(r"[^\d]","",regex = True)
df1['rating_count'] = df1['rating_count'].astype(float)
```

```python
df1['rating_count'] = df1['rating_count'].astype(int)
```

```python
# COnverting the str to float for remaining
df1['discounted_price'].value_counts()
```

```python
df1["discounted_price"] = df1["discounted_price"].str.replace(r"[^\d]", "", regex=True)
df1['discounted_price'] = df1['discounted_price'].astype(float)
```

```python
df1['actual_price'].value_counts()
```

```python
df1["actual_price"] = df1["actual_price"].str.replace(r"[^\d]", "", regex=True)
df1['actual_price'] = df1['actual_price'].astype(float)
```

```python
df1["discount_percentage"] = df1["discount_percentage"].str.replace(r"[^\d]", "", regex=True)
df1['discount_percentage'] = df1['discount_percentage'].astype(float)
df1['discount_percentage'] = df1['discount_percentage']/100
```
